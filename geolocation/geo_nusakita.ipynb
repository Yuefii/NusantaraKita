```python
import pymysql
import json
import os

# koneksi MySQL
conn = pymysql.connect(
    host='localhost',
    user='root',
    password='',
    database='geolocation',
    charset='utf8mb4'
)

cursor = conn.cursor()
print("✅ Connected to MySQL")
```

```python
base_path = r"C:\geolocation\output"

folders = [
    "provinsi",
    "kabupaten_kota",
    "kecamatan",
    "desa_kelurahan"
]

for f in folders:
    path = os.path.join(base_path, f)
    os.makedirs(path, exist_ok=True)

print("✅ Folder siap:", folders)
```

```python
def swap_coordinates(coords):
    """
    Membalik [lat, lon] menjadi [lon, lat]
    dan mempertahankan nested arrays.
    """
    if isinstance(coords, list):
        # Jika elemen pertama adalah angka → [lat, lon]
        if len(coords) == 2 and all(isinstance(x, (int, float)) for x in coords):
            return [coords[1], coords[0]]  # swap
        else:
            return [swap_coordinates(c) for c in coords]
    return coords
```

```python
def save_geojson(output_file, path_data):
    geojson = {
        "type": "MultiPolygon",
        "coordinates": swap_coordinates(path_data)
    }

    with open(output_file, "w", encoding="utf-8") as f:
        json.dump(geojson, f, indent=2)

    print("✅ Saved:", output_file)
```

```python
cursor.execute("SELECT kode, path FROM geo_provinsi")

for kode, path in cursor.fetchall():
    coords = json.loads(path)
    file_path = os.path.join(base_path, "provinsi", f"{kode}.geojson")
    save_geojson(file_path, coords)
```

```python
cursor.execute("SELECT kode, path FROM geo_kabupaten_kota")

for kode, path in cursor.fetchall():
    coords = json.loads(path)
    file_path = os.path.join(base_path, "kabupaten_kota", f"{kode}.geojson")
    save_geojson(file_path, coords)
```

```python
cursor.execute("SELECT kode, path FROM geo_kecamatan")

for kode, path in cursor.fetchall():
    coords = json.loads(path)
    file_path = os.path.join(base_path, "kecamatan", f"{kode}.geojson")
    save_geojson(file_path, coords)
```

```python
cursor.execute("SELECT kode, path FROM geo_desa_kelurahan")

for kode, path in cursor.fetchall():
    coords = json.loads(path)
    file_path = os.path.join(base_path, "desa_kelurahan", f"{kode}.geojson")
    save_geojson(file_path, coords)
```

```python
cursor.close()
conn.close()
print("✅ DONE — Semua GeoJSON berhasil dibuat")
```
